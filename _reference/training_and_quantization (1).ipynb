````python
import torch
import torch.nn as nn

# -----------------------------
# 1. Exact model from author's notebook
# -----------------------------

class DoppelKonv(nn.Module):
    """Zwei aufeinanderfolgende Faltungen"""
    def __init__(self, in_channels, out_channels):
        super().__init__()

        self.conv = nn.Sequential(
            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size=3,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),

            nn.Conv2d(
                out_channels,
                out_channels,
                kernel_size=3,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.conv(x)


class UNet2D(nn.Module):

    def __init__(self, in_channels=1, out_channels=1):
        super().__init__()

        # Encoder
        self.down1 = DoppelKonv(in_channels, 32)
        self.pool1 = nn.MaxPool2d(kernel_size=2, stride=2)

        self.down2 = DoppelKonv(32, 64)
        self.pool2 = nn.MaxPool2d(kernel_size=2, stride=2)

        self.down3 = DoppelKonv(64, 128)
        self.pool3 = nn.MaxPool2d(kernel_size=2, stride=2)

        # Bridge
        self.bridge = DoppelKonv(128, 256)

        # Decoder
        self.up3 = nn.ConvTranspose2d(
            256, 128, kernel_size=2, stride=2
        )
        self.up_conv3 = DoppelKonv(256, 128)

        self.up2 = nn.ConvTranspose2d(
            128, 64, kernel_size=2, stride=2
        )
        self.up_conv2 = DoppelKonv(128, 64)

        self.up1 = nn.ConvTranspose2d(
            64, 32, kernel_size=2, stride=2
        )
        self.up_conv1 = DoppelKonv(64, 32)

        # Output
        self.final_conv = nn.Conv2d(
            32, out_channels, kernel_size=1
        )

    def forward(self, x):

        # Encoder
        x1 = self.down1(x)
        x2 = self.down2(self.pool1(x1))
        x3 = self.down3(self.pool2(x2))

        # Bridge
        b = self.bridge(self.pool3(x3))

        # Decoder + skip connections
        u3 = self.up3(b)
        u3 = torch.cat([u3, x3], dim=1)
        u3 = self.up_conv3(u3)

        u2 = self.up2(u3)
        u2 = torch.cat([u2, x2], dim=1)
        u2 = self.up_conv2(u2)

        u1 = self.up1(u2)
        u1 = torch.cat([u1, x1], dim=1)
        u1 = self.up_conv1(u1)

        return self.final_conv(u1)


# -----------------------------
# 2. Create model
# -----------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = UNet2D().to(device)


# -----------------------------
# 3. Load trained weights
# -----------------------------

weights_path = "/content/brats_unet_modell.pth"

state_dict = torch.load(
    weights_path,
    map_location=device
)

model.load_state_dict(state_dict)

model.eval()


# -----------------------------
# 4. Verify
# -----------------------------

print("✅ Model created")
print("✅ Weights loaded")
print("Device:", device)
````

````python
import torch
import torch.nn as nn

# -----------------------------
# 1. Define the exact U-Net
# -----------------------------

class DoppelKonv(nn.Module):
    def __init__(self, in_channels, out_channels):
        super().__init__()

        self.conv = nn.Sequential(
            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size=3,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),

            nn.Conv2d(
                out_channels,
                out_channels,
                kernel_size=3,
                padding=1,
                bias=False
            ),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True)
        )

    def forward(self, x):
        return self.conv(x)


class UNet2D(nn.Module):
    def __init__(self, in_channels=1, out_channels=1):
        super().__init__()

        # Encoder
        self.down1 = DoppelKonv(in_channels, 32)
        self.pool1 = nn.MaxPool2d(2, 2)

        self.down2 = DoppelKonv(32, 64)
        self.pool2 = nn.MaxPool2d(2, 2)

        self.down3 = DoppelKonv(64, 128)
        self.pool3 = nn.MaxPool2d(2, 2)

        # Bridge
        self.bridge = DoppelKonv(128, 256)

        # Decoder
        self.up3 = nn.ConvTranspose2d(
            256, 128, kernel_size=2, stride=2
        )
        self.up_conv3 = DoppelKonv(256, 128)

        self.up2 = nn.ConvTranspose2d(
            128, 64, kernel_size=2, stride=2
        )
        self.up_conv2 = DoppelKonv(128, 64)

        self.up1 = nn.ConvTranspose2d(
            64, 32, kernel_size=2, stride=2
        )
        self.up_conv1 = DoppelKonv(64, 32)

        # Output
        self.final_conv = nn.Conv2d(
            32, out_channels, kernel_size=1
        )

    def forward(self, x):

        x1 = self.down1(x)
        x2 = self.down2(self.pool1(x1))
        x3 = self.down3(self.pool2(x2))

        b = self.bridge(self.pool3(x3))

        u3 = self.up3(b)
        u3 = torch.cat([u3, x3], dim=1)
        u3 = self.up_conv3(u3)

        u2 = self.up2(u3)
        u2 = torch.cat([u2, x2], dim=1)
        u2 = self.up_conv2(u2)

        u1 = self.up1(u2)
        u1 = torch.cat([u1, x1], dim=1)
        u1 = self.up_conv1(u1)

        return self.final_conv(u1)


# -----------------------------
# 2. Create model
# -----------------------------

device = torch.device("cpu")

model = UNet2D().to(device)

# -----------------------------
# 3. Load trained weights
# -----------------------------

weights_path = "/content/brats_unet_modell.pth"

state_dict = torch.load(
    weights_path,
    map_location=device
)

model.load_state_dict(state_dict)
model.eval()

print("✅ Model recreated")
print("✅ Trained weights loaded")
print("Device:", device)

# -----------------------------
# 4. Quick architecture check
# -----------------------------

dummy = torch.randn(1, 1, 240, 240)

with torch.no_grad():
    output = model(dummy)

print("Input :", dummy.shape)
print("Output:", output.shape)
````

````python
from onnxruntime.quantization import (
    CalibrationDataReader,
    QuantType,
    QuantFormat,
    quantize_static
)

class MRIDataReader(CalibrationDataReader):

    def __init__(self, data, input_name):
        self.data = data
        self.input_name = input_name
        self.index = 0

    def get_next(self):
        if self.index >= len(self.data):
            return None

        sample = {
            self.input_name: self.data[self.index:self.index + 1]
        }

        self.index += 1
        return sample


# Create calibration reader
calibration_reader = MRIDataReader(
    calibration_data,
    session.get_inputs()[0].name
)

fp32_model = "/content/ieee_brain_tumor/unet_fp32.onnx"
int8_model = "/content/ieee_brain_tumor/unet_int8.onnx"

print("Starting static INT8 quantization...")

quantize_static(
    model_input=fp32_model,
    model_output=int8_model,
    calibration_data_reader=calibration_reader,
    quant_format=QuantFormat.QDQ,
    activation_type=QuantType.QInt8,
    weight_type=QuantType.QInt8,
    per_channel=True
)

print("\n✅ INT8 quantization complete!")
````

````python
import time
import numpy as np

# Same real MRI input for both
benchmark_input = input_tensor_real.numpy().astype(np.float32)

fp32_input_name = session.get_inputs()[0].name
int8_input_name = int8_session.get_inputs()[0].name

def benchmark_onnx(sess, input_name, x, warmup=20, runs=200):

    feed = {input_name: x}

    # Warm-up
    for _ in range(warmup):
        sess.run(None, feed)

    times = []

    for _ in range(runs):
        start = time.perf_counter()
        sess.run(None, feed)
        end = time.perf_counter()

        times.append((end - start) * 1000)

    return np.mean(times), np.std(times), np.median(times)


fp32_mean, fp32_std, fp32_median = benchmark_onnx(
    session,
    fp32_input_name,
    benchmark_input
)

int8_mean, int8_std, int8_median = benchmark_onnx(
    int8_session,
    int8_input_name,
    benchmark_input
)

print("========== FAIR CPU BENCHMARK ==========")

print(
    f"FP32 → Mean: {fp32_mean:.2f} ms | "
    f"Std: {fp32_std:.2f} | "
    f"Median: {fp32_median:.2f} ms"
)

print(
    f"INT8 → Mean: {int8_mean:.2f} ms | "
    f"Std: {int8_std:.2f} | "
    f"Median: {int8_median:.2f} ms"
)

print(
    f"\nSpeedup: {fp32_mean / int8_mean:.2f}x"
)

print(
    f"Latency reduction: "
    f"{(1 - int8_mean / fp32_mean) * 100:.2f}%"
)
````

````python
# Full validation metrics: FP32 vs INT8
# Metrics: Dice, IoU, Precision, Recall

import h5py
import numpy as np

fp32_dice = []
fp32_iou = []
fp32_precision = []
fp32_recall = []

int8_dice = []
int8_iou = []
int8_precision = []
int8_recall = []

fp32_input_name = session.get_inputs()[0].name
int8_input_name = int8_session.get_inputs()[0].name


def calculate_metrics(pred, true):

    pred = pred.astype(bool)
    true = true.astype(bool)

    tp = np.logical_and(pred, true).sum()
    fp = np.logical_and(pred, ~true).sum()
    fn = np.logical_and(~pred, true).sum()

    dice = (2 * tp) / (2 * tp + fp + fn + 1e-8)

    iou = tp / (tp + fp + fn + 1e-8)

    precision = tp / (tp + fp + 1e-8)

    recall = tp / (tp + fn + 1e-8)

    return dice, iou, precision, recall


for i, file_path in enumerate(val_files):

    with h5py.File(file_path, "r") as f:
        image = f["x"][:]
        mask = f["y"][:]

    x = image.astype(np.float32)[None, None, :, :]
    true_binary = (mask > 0).astype(np.uint8)

    # ---------------- FP32 ----------------
    fp32_output = session.run(
        None,
        {fp32_input_name: x}
    )[0]

    fp32_probability = 1 / (1 + np.exp(-fp32_output))

    fp32_prediction = (
        fp32_probability > 0.5
    ).astype(np.uint8)[0, 0]

    metrics = calculate_metrics(
        fp32_prediction,
        true_binary
    )

    fp32_dice.append(metrics[0])
    fp32_iou.append(metrics[1])
    fp32_precision.append(metrics[2])
    fp32_recall.append(metrics[3])

    # ---------------- INT8 ----------------
    int8_output = int8_session.run(
        None,
        {int8_input_name: x}
    )[0]

    int8_probability = 1 / (1 + np.exp(-int8_output))

    int8_prediction = (
        int8_probability > 0.5
    ).astype(np.uint8)[0, 0]

    metrics = calculate_metrics(
        int8_prediction,
        true_binary
    )

    int8_dice.append(metrics[0])
    int8_iou.append(metrics[1])
    int8_precision.append(metrics[2])
    int8_recall.append(metrics[3])

    if (i + 1) % 100 == 0:
        print(f"Processed {i + 1}/{len(val_files)}")


print("\n========== FULL VALIDATION METRICS ==========")

print("\nFP32:")
print(f"Dice:      {np.mean(fp32_dice) * 100:.2f}%")
print(f"IoU:       {np.mean(fp32_iou) * 100:.2f}%")
print(f"Precision: {np.mean(fp32_precision) * 100:.2f}%")
print(f"Recall:    {np.mean(fp32_recall) * 100:.2f}%")

print("\nINT8:")
print(f"Dice:      {np.mean(int8_dice) * 100:.2f}%")
print(f"IoU:       {np.mean(int8_iou) * 100:.2f}%")
print(f"Precision: {np.mean(int8_precision) * 100:.2f}%")
print(f"Recall:    {np.mean(int8_recall) * 100:.2f}%")
````

````python
import os

inference_code = r'''
import argparse
import h5py
import numpy as np
import onnxruntime as ort
import matplotlib.pyplot as plt


def load_mri(h5_path):
    """Load one MRI slice from an HDF5 file."""
    with h5py.File(h5_path, "r") as f:
        image = f["x"][:]

    return image


def preprocess(image):
    """Convert MRI slice to ONNX input format."""
    return image.astype(np.float32)[None, None, :, :]


def predict(session, image):
    """Run INT8 ONNX inference."""
    input_name = session.get_inputs()[0].name

    output = session.run(
        None,
        {input_name: preprocess(image)}
    )[0]

    # Convert logits to probability
    probability = 1.0 / (1.0 + np.exp(-output))

    # Binary tumor mask
    mask = (probability > 0.5).astype(np.uint8)[0, 0]

    return probability[0, 0], mask


def save_result(image, mask, output_path):
    """Save MRI + predicted tumor mask."""
    plt.figure(figsize=(10, 4))

    plt.subplot(1, 2, 1)
    plt.imshow(image, cmap="gray")
    plt.title("Input MRI")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(mask, cmap="gray")
    plt.title("Predicted Tumor Mask")
    plt.axis("off")

    plt.tight_layout()
    plt.savefig(output_path, dpi=200, bbox_inches="tight")
    plt.close()


def main():

    parser = argparse.ArgumentParser(
        description="Brain Tumor Segmentation using INT8 ONNX U-Net"
    )

    parser.add_argument(
        "--model",
        default="models/unet_int8.onnx",
        help="Path to INT8 ONNX model"
    )

    parser.add_argument(
        "--input",
        required=True,
        help="Path to MRI .h5 file"
    )

    parser.add_argument(
        "--output",
        default="prediction.png",
        help="Output visualization path"
    )

    args = parser.parse_args()

    print("Loading INT8 ONNX model...")

    session = ort.InferenceSession(
        args.model,
        providers=["CPUExecutionProvider"]
    )

    print("Model loaded successfully.")

    image = load_mri(args.input)

    print("MRI shape:", image.shape)
    print("MRI dtype:", image.dtype)

    probability, mask = predict(
        session,
        image
    )

    print("Prediction completed.")
    print("Predicted tumor pixels:", int(mask.sum()))

    save_result(
        image,
        mask,
        args.output
    )

    print("Result saved:", args.output)


if __name__ == "__main__":
    main()
'''

inference_path = "/content/ieee_brain_tumor/inference.py"

with open(inference_path, "w") as f:
    f.write(inference_code)

print("✅ inference.py created")
print("Path:", inference_path)
````

````python
import os

inference_code = r'''
import argparse
import h5py
import numpy as np
import onnxruntime as ort
import matplotlib.pyplot as plt


def load_mri_and_mask(h5_path):
    """Load MRI slice and ground-truth mask."""
    with h5py.File(h5_path, "r") as f:
        image = f["x"][:]
        mask = f["y"][:]

    return image, mask


def preprocess(image):
    """Convert MRI to ONNX input format."""
    return image.astype(np.float32)[None, None, :, :]


def predict(session, image):
    """Run INT8 ONNX inference."""
    input_name = session.get_inputs()[0].name

    output = session.run(
        None,
        {input_name: preprocess(image)}
    )[0]

    probability = 1.0 / (1.0 + np.exp(-output))

    prediction = (
        probability > 0.5
    ).astype(np.uint8)[0, 0]

    return prediction


def calculate_dice(prediction, ground_truth):
    """Calculate binary Dice score."""

    prediction = prediction.astype(bool)
    ground_truth = ground_truth.astype(bool)

    intersection = np.logical_and(
        prediction,
        ground_truth
    ).sum()

    dice = (
        2.0 * intersection
        / (
            prediction.sum()
            + ground_truth.sum()
            + 1e-8
        )
    )

    return dice


def save_result(
    image,
    ground_truth,
    prediction,
    dice,
    output_path
):
    """Save MRI, ground truth and prediction comparison."""

    ground_truth_binary = (ground_truth > 0).astype(np.uint8)

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(image, cmap="gray")
    plt.title("MRI")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(ground_truth_binary, cmap="gray")
    plt.title("Ground Truth")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(prediction, cmap="gray")
    plt.title(f"INT8 Prediction — Dice {dice * 100:.2f}%")
    plt.axis("off")

    plt.tight_layout()
    plt.savefig(
        output_path,
        dpi=200,
        bbox_inches="tight"
    )
    plt.close()


def main():

    parser = argparse.ArgumentParser(
        description="Brain Tumor Segmentation using INT8 ONNX U-Net"
    )

    parser.add_argument(
        "--model",
        default="models/unet_int8.onnx"
    )

    parser.add_argument(
        "--input",
        required=True
    )

    parser.add_argument(
        "--output",
        default="prediction.png"
    )

    args = parser.parse_args()

    print("Loading INT8 ONNX model...")

    session = ort.InferenceSession(
        args.model,
        providers=["CPUExecutionProvider"]
    )

    print("Model loaded successfully.")

    image, ground_truth = load_mri_and_mask(
        args.input
    )

    print("MRI shape:", image.shape)
    print("MRI dtype:", image.dtype)

    prediction = predict(
        session,
        image
    )

    ground_truth_binary = (
        ground_truth > 0
    ).astype(np.uint8)

    dice = calculate_dice(
        prediction,
        ground_truth_binary
    )

    print("Prediction completed.")
    print("Predicted tumor pixels:", int(prediction.sum()))
    print("Ground-truth tumor pixels:", int(ground_truth_binary.sum()))
    print(f"Dice score: {dice * 100:.2f}%")

    save_result(
        image,
        ground_truth,
        prediction,
        dice,
        args.output
    )

    print("Result saved:", args.output)


if __name__ == "__main__":
    main()
'''

inference_path = "/content/ieee_brain_tumor/inference.py"

with open(inference_path, "w") as f:
    f.write(inference_code)

print("✅ inference.py updated")
print("Now includes:")
print("  • MRI")
print("  • Ground Truth")
print("  • INT8 Prediction")
print("  • Dice score")
````
